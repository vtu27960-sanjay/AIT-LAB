```python
operators = {

    'walk': {
        'pre': {'on-floor(monkey)', 'at(monkey, center)'},
        'add': {'at(monkey, window)'},
        'del': {'at(monkey, center)'}
    },

    'push': {
        'pre': {'on-floor(monkey)', 'at(monkey, window)',
                'at(box, window)'},
        'add': {'at(monkey, center)', 'at(box, center)'},
        'del': {'at(monkey, window)', 'at(box, window)'}
    },

    'climb': {
        'pre': {'on-floor(monkey)', 'at(monkey, center)',
                'at(box, center)'},
        'add': {'on-box(monkey)'},
        'del': {'on-floor(monkey)'}
    },

    'grab': {
        'pre': {'on-box(monkey)', 'at(monkey, center)',
                'at(box, center)', 'at(banana, center)'},
        'add': {'has(monkey, banana)'},
        'del': set()
    }
}


state = {
    'on-floor(monkey)',
    'at(monkey, center)',
    'at(box, window)',
    'at(banana, center)'
}

goal = {'has(monkey, banana)'}


for action, op in operators.items():

    if op['pre'].issubset(state):

        state -= op['del']
        state |= op['add']

        print(action)

        if goal.issubset(state):
            break


print("Final state:", state)
```
