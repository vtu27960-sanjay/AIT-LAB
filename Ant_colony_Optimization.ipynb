```python
import matplotlib.pyplot as plt

R1 = 30
R2 = 70

total = R1 + R2

p1 = R1 / total
p2 = R2 / total

percentage1 = p1 * 100
percentage2 = p2 * 100

print("Path 1 probability:", percentage1, "%")
print("Path 2 probability:", percentage2, "%")

if p2 > p1:
    print("\nResults:")
    print("Most ants follow path 2 because it has  a 70% chance.")

elif p1 > p2:
    print("\nResults:")
    print("Most ants follow path 1 because it has a higher probability.")

else:
    print("\nResults:")
    print("Both paths have the same probability.")

plt.bar(["Path 1", "Path 2"], [percentage1, percentage2])
plt.show()
```
