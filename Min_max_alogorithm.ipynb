```python
height = int(input("Enter tree height: "))

n = 2 ** height

print("Enter", n, "leaf node values:")
values = list(map(int, input().split()))

if len(values) != n:
    print("Please enter exactly", n, "values.")
else:
    print("Leaf values:", values)

    level = 0

    while len(values) > 1:
        new_values = []

        for i in range(0, len(values), 2):
            if level % 2 == 0:
                new_values.append(max(values[i], values[i + 1]))
            else:
                new_values.append(min(values[i], values[i + 1]))

        values = new_values
        level += 1

    print("Minimax result:", values[0])
```
